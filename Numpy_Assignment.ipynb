```python
import numpy as np

expenses = np.array([1500, 2000, 1800, 2200, 1700, 2500])

#1. Average monthly expense
#To find the average, we use the function mean()
average_expenses = np.mean(expenses)
print(f"The Average monthly expense is: {np.mean(expenses)}")

#2. Highest expense month
#To find the highest number, we use the function max()
highest_expenses = np.max(expenses)
print(f"The Highest expense month is: {highest_expenses}")

#3. Lowest expense month
#To find the lowest, we use the function min()
lowest_expenses = np.min(expenses)
print(f"The Lowest Expense month is:{np.min(expenses)}")

#4. Total expenses
#To find the total ot the sum, we use the function sum()
total_expenses = expenses.sum()
print(f"The Total expenses are: {expenses.sum()}")

#5. Slice to show only the first 3 months
#To slice the first 3 months, we start with index [0] since that's the first number, then we end right before the intended number which idealy is 2, but we're assigning [3] because 2 is the number before it.
print(f"The first three monthly expenses are: {expenses[0:3]}")
```
