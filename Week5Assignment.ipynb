```python
import pandas as pd
import numpy as np
data = {
'Order_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
'Customer_Name': ['Amara', 'Bola', None, 'Diana', 'Efua', 'Fatima', 'Grace',
None, 'Ifeoma', 'Joyce'],
'Country': ['Ghana', 'Nigeria', 'Kenya', None, 'Ghana', 'Cameroon', None,
'Kenya', 'Nigeria', 'Ghana'],
'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', None, 'Phone', 'Tablet',
'Laptop', None, 'Phone'],
'Quantity': [2, 1, None, 3, 2, None, 1, 4, 2, None],
'Price': [1500, 800, 600, 1500, None, 800, 600, 1500, None, 800],
'Delivery_Days': [3, None, 7, 5, 3, None, 6, None, 4, 5]
}
df = pd.DataFrame(data)
print(df)
```

```python
#TASK 1 - Explore the data

#The dataset contains 10 orders across the 7 columns, with a mix of text and numeric data types. 
#Running df.isnull().sum() reveals that 6 out of 7 columns have at least one missing value, with a total of 10 missing entries across the entire dataset.
#The affected columns are: Customer_Name (2 missing), Country (2), Product (2), Quantity (3), Price (2), and Delivery_Days (3).
#Only Order_ID has no missing values, making it the only column we can fully rely on to identify each record.
```

```python
#TASK 2 - Handle missing customer names
df['Customer_Name'] = df['Customer_Name'].fillna('Unknown')
print(df[['Order_ID', 'Customer_Name']])



#Since we cannot recover the actual customer names, "Unknown" is an honest placeholder that keeps the data usable without misrepresenting who placed the order.
```

```python
#TASK 3- Handle missing prices and quantity

#For price:
price_median = df['Price'].median()
df['Price'] = df['Price'].fillna(price_median)
print(f"Price filled with median: {price_median}")

#I chose the MEDIAN to fill missing Price values because the Price column contains very different values (600, 800, 1500) 
#thisspread means the mean would be pulled toward the higher prices and misrepresent a typical order. 
#The median (800) reflects the middle of the actual distribution more accurately.
```

```python
#For product
quantity_median = df['Quantity'].median()
df['Quantity'] = df['Quantity'].fillna(quantity_median)
print(f"Quantity filled with median: {quantity_median}")

#Using the mean (2.0 in this case) would give the same result, but the median is the safer habit for small datasets where outliers can have a big impact.
```

```python
#TASK 4 — HANDLE MISSING DELIVERY DAYS

#Check for outliers using IQR method
delivery = df['Delivery_Days'].dropna()
print("Delivery_Days values:", delivery.tolist())
print("Mean:  ", delivery.mean())
print("Median:", delivery.median())
```

```python
Q1 = delivery.quantile(0.25)
Q3 = delivery.quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
 
print(f"Q1={Q1}, Q3={Q3}, IQR={IQR}")
print(f"Outlier bounds: lower={lower_bound}, upper={upper_bound}")
 
outliers = delivery[(delivery < lower_bound) | (delivery > upper_bound)]
print(f"Outliers found: {outliers.tolist()}")
```

```python
delivery_mean = (df['Delivery_Days'].mean())
df['Delivery_Days'] = df['Delivery_Days'].fillna(delivery_mean)
 
print(f"Delivery_Days filled with mean: {delivery_mean}")
print("Updated Delivery_Days column:")
print(df[['Order_ID', 'Delivery_Days']])
```

```python
#Using the IQR method, I checked for outliers in Delivery_Days.
#The values (3, 7, 5, 3, 6, 4, 5) fall within a range with no outliers detected — all values sit comfortably between the lower and upper bounds. 
#Because the data is clean and evenly distributed, the mean (4.7 days) is a fair and accurate representation of a typical delivery time, making it the right choice to fill the three missing values.
```

```python
# TASK 5 — HANDLE MISSING COUNTRY AND PRODUCT

#For country:
country_mode = df['Country'].mode()[0]
df['Country'] = df['Country'].fillna(country_mode)
print(f"Country filled with mode: '{country_mode}'")
```

```python
#For product:
product_mode = df['Product'].mode()[0]
df['Product'] = df['Product'].fillna(product_mode)
print(f"Product filled with mode: '{product_mode}'")
```

```python
#I chose to fill in the missing Country and Product values with the mode because dropping rows with missing products would discard orders that still have valid customer, price, and delivery data.
```

```python
#TASK 6 — CONFIRM YOUR WORK
print(df.isnull().sum())
```
