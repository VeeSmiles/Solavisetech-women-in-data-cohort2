```python
import pandas as pd

data = {
    'Order_ID': ['ORD001','ORD002','ORD003','ORD004','ORD005',
                 'ORD006','ORD007','ORD008','ORD009','ORD010'],
    'Product': ['Laptop','Rice Bag','Headphones','Notebook','Blender',
                'Rice Bag','Laptop','Notebook','Blender','Headphones'],
    'Category': ['Electronics','Food','Electronics','Stationery','Appliances',
                 'Food','Electronics','Stationery','Appliances','Electronics'],
    'Region': ['North','South','East','West','North',
               'East','South','North','West','North'],
    'Units_Sold': [5, 30, 12, 50, 8, 20, 3, 80, 6, 15],
    'Unit_Price': [850.00, 12.50, 45.00, 3.20, 65.00,
                   12.50, 850.00, 3.20, 65.00, 45.00],
    'Month': ['January','January','February','February','March',
              'March','April','April','May','May']
}

df = pd.DataFrame(data)
print(df)
```

```python
#1. Display only the Product and Region columns.
print(df[['Product', 'Region']])
```

```python
#2. Display only the Order_ID, Category and Units_Sold columns.
df[['Order_ID', 'Category', 'Units_Sold']]
```

```python
#3. Show the Unit_Price column using a single bracket. Then show it again using double brackets. 
#What is different about the output of each one?

# single bracket
df['Unit_Price']
```

```python
# double brackets
df[['Unit_Price']]  
```

```python
# Single bracket - returns a Series (one-dimensional, just an index and values).
# Double brackets - returns a DataFrame (two-dimensional, keeps the column header/table structure).
```

```python
#4. Show only the rows where Region is North.
df[df['Region'] == 'North']
```

```python
#5. Show only the rows where Units_Sold is greater than 10.
df[df['Units_Sold'] > 10]
```

```python
#6. Show only the rows where Category is Electronics AND Units_Sold is greater than 5.
df[(df['Category'] == 'Electronics') & (df['Units_Sold'] > 5)]
```

```python
#7. Show only the rows where Region is South OR Region is West.
df[(df['Region'] == 'South') | (df['Region'] == 'West')]
```

```python
#8. Find the total Units_Sold for each Region.
df.groupby('Region')['Units_Sold'].sum()
```

```python
#9. Find the average Unit_Price for each Category.
df.groupby('Category')['Unit_Price'].mean()
```

```python
#10. Find the total Units_Sold AND the average Unit_Price for each Category in one step.
#(Hint: Use .agg())
df.groupby('Category').agg({'Units_Sold': 'sum', 'Unit_Price': 'mean'})
```

```python
#11. Find the total Units_Sold for each Month.
df.groupby('Month')['Units_Sold'].sum()
```

```python
#12. Which region had the highest total Units_Sold?
#(Hint: Use groupby, then sum, then idxmax)
df.groupby('Region')['Units_Sold'].sum().idxmax()
```

```python
#13. Which product had the lowest average Unit_Price?
#(Hint: Use groupby, then mean, then idxmin)
df.groupby('Product')['Unit_Price'].mean().idxmin()
```

```python
#14. Which month had the highest total Units_Sold?
df.groupby('Month')['Units_Sold'].sum().idxmax()
```

```python
#15. Show only the Electronics orders from the North region. Display only the Product, Units_Sold and Unit_Price columns.
result = df[(df['Category'] == 'Electronics') & (df['Region'] == 'North')]
print(result[['Product', 'Units_Sold', 'Unit_Price']])
```

```python
#16. Create a new column called Total_Sales by multiplying Units_Sold by Unit_Price. 
#Then find which category had the highest total sales overall.
df['Total_Sales'] = df['Units_Sold'] * df['Unit_Price']
print(df.groupby('Category')['Total_Sales'].sum().idxmax())
```

```python
# For idxmax()/idxmin(), these return the label (e.g. the Region or Product name) corresponding to the max/min value of a grouped Series.
# That's why they're hinted for questions 12, 13, and 14, since the question asks "which" one, not "what" the value is.
```
